```python
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.animation as animation


plt.rcParams['font.family'] = 'NanumBarunGothic'
plt.rcParams['axes.unicode_minus'] = False
stores = [
    "빠오즈푸", "베키아에누보", "이쿠", "봄이보리밥",
    "마복림할머니집", "운치", "트라토리아진", "낙지일번지",
    "다온카츠", "샤브인서울", "익선취향", "이태리맨션",
    "땀땀", "을밀대 평양냉면", "담금", "곱창컴퍼니"
]

edges = [
    ("빠오즈푸", "베키아에누보", 15), ("빠오즈푸", "이쿠", 20),
    ("베키아에누보", "봄이보리밥", 10), ("베키아에누보", "마복림할머니집", 25),
    ("이쿠", "운치", 30), ("이쿠", "트라토리아진", 12),
    ("봄이보리밥", "낙지일번지", 18), ("마복림할머니집", "다온카츠", 22),
    ("운치", "샤브인서울", 15), ("트라토리아진", "익선취향",  12),
    ("낙지일번지", "이태리맨션", 25), ("다온카츠", "땀땀", 14),
    ("샤브인서울", "을밀대 평양냉면", 19), ("익선취향", "담금", 8),
    ("이태리맨션", "곱창컴퍼니", 35), ("땀땀", "을밀대 평양냉면", 11),
    ("담금", "곱창컴퍼니", 17), ("빠오즈푸", "마복림할머니집", 40)
]


G = nx.Graph()
G.add_nodes_from(stores)
G.add_weighted_edges_from(edges)

pos = nx.spring_layout(G, seed=42)



# 2) DFS
def get_dfs_order(graph, start_node):
    visited = []
    def dfs(node):
        if node not in visited:
            visited.append(node)
            for neighbor in sorted(graph.neighbors(node)):
                dfs(neighbor)
    dfs(start_node)
    return visited

# 3) BFS
def get_bfs_order(graph, start_node):
    visited = [start_node]
    queue = [start_node]
    while queue:
        curr = queue.pop(0)
        for neighbor in sorted(graph.neighbors(curr)):
            if neighbor not in visited:
                visited.append(neighbor)
                queue.append(neighbor)
    return visited

# 4) 프림 (Prim) 알고리즘
def get_prim_edges(graph, start_node):
    mst_edges = []
    visited = {start_node}

    while len(visited) < len(graph.nodes()):
        possible_edges = []
        for u in visited:
            for v in graph.neighbors(u):
                if v not in visited:
                    weight = graph[u][v]['weight']
                    possible_edges.append((weight, u, v))

        if not possible_edges:
            break


        possible_edges.sort(key=lambda x: (x[0], x[1], x[2]))
        weight, u, v = possible_edges[0]

        mst_edges.append((u, v))
        visited.add(v)

    return mst_edges

# 5) 크루스칼 (Kruskal) 알고리즘
def get_kruskal_edges(graph):
    mst_edges = []


    parent = {node: node for node in graph.nodes()}

    def find(node):
        if parent[node] == node:
            return node
        parent[node] = find(parent[node])
        return parent[node]

    def union(node1, node2):
        root1 = find(node1)
        root2 = find(node2)
        if root1 != root2:
            parent[root2] = root1
            return True
        return False

    all_edges = []
    for u, v, data in graph.edges(data=True):
        all_edges.append((data['weight'], u, v))
    all_edges.sort(key=lambda x: x[0])

    for weight, u, v in all_edges:
        if union(u, v):
            mst_edges.append((u, v))
            if len(mst_edges) == len(graph.nodes()) - 1:
                break

    return mst_edges



start_node = stores[0]
dfs_order = get_dfs_order(G, start_node)
bfs_order = get_bfs_order(G, start_node)
prim_history = get_prim_edges(G, start_node)
kruskal_history = get_kruskal_edges(G)


# 2, 3번용: 노드 방문 애니메이션
def build_animation(graph, pos, visit_order, filename, title_name):
    fig, ax = plt.subplots(figsize=(12, 8))
    edge_labels = nx.get_edge_attributes(graph, 'weight')

    def update(frame):
        ax.clear()
        ax.set_title(f"{title_name} 탐색 과정 (프레임: {frame+1}/{len(visit_order)})", fontsize=16, fontweight='bold')

        current_visited = visit_order[:frame+1]
        node_colors = []
        for node in graph.nodes():
            if node == visit_order[frame]:
                node_colors.append('red')
            elif node in current_visited:
                node_colors.append('orange')
            else:
                node_colors.append('skyblue')

        nx.draw_networkx_nodes(graph, pos, node_color=node_colors, node_size=700, ax=ax)
        nx.draw_networkx_edges(graph, pos, width=1.5, alpha=0.6, edge_color='gray', ax=ax)
        nx.draw_networkx_labels(graph, pos, font_size=10, font_family='NanumBarunGothic', ax=ax)
        nx.draw_networkx_edge_labels(graph, pos, edge_labels=edge_labels, font_size=8, ax=ax)
        ax.axis('off')

    ani = animation.FuncAnimation(fig, update, frames=len(visit_order), repeat=False, interval=1000)

    print(f"{filename} 저장 중...")
    try:
        ani.save(filename, writer='ffmpeg', fps=1)
        print(f"성공적으로 저장되었습니다: {filename}")
    except Exception as e:
        print(f"mp4 저장 실패: {e}. 대안으로 GIF를 시도합니다.")
        ani.save(filename.replace('.mp4', '.gif'), writer='pillow', fps=1)
    plt.close(fig)


# 4, 5번용: MST(최소 신장 트리) 간선 추가 애니메이션
def build_mst_animation(graph, pos, edge_history, filename, title_name):

    total_frames = len(edge_history) + 1
    fig, ax = plt.subplots(figsize=(12, 8))
    edge_labels = nx.get_edge_attributes(graph, 'weight')

    def update(frame):
        ax.clear()
        ax.set_title(f"{title_name} 생성 과정 (단계: {frame}/{len(edge_history)})", fontsize=16, fontweight='bold')


        current_mst_edges = edge_history[:frame]


        edge_colors = []
        edge_widths = []


        mst_set = {tuple(sorted(e)) for e in current_mst_edges}

        for u, v in graph.edges():
            if tuple(sorted((u, v))) in mst_set:

                if frame > 0 and tuple(sorted((u, v))) == tuple(sorted(edge_history[frame-1])):
                    edge_colors.append('red')
                    edge_widths.append(4.0)
                else:
                    edge_colors.append('orange')
                    edge_widths.append(3.5)
            else:
                edge_colors.append('lightgray')
                edge_widths.append(1.0)


        mst_nodes = set()
        for u, v in current_mst_edges:
            mst_nodes.add(u)
            mst_nodes.add(v)

        node_colors = ['orange' if node in mst_nodes else 'skyblue' for node in graph.nodes()]

        nx.draw_networkx_nodes(graph, pos, node_color=node_colors, node_size=700, ax=ax)
        nx.draw_networkx_edges(graph, pos, width=edge_widths, edge_color=edge_colors, ax=ax)
        nx.draw_networkx_labels(graph, pos, font_size=10, font_family='NanumBarunGothic', ax=ax)
        nx.draw_networkx_edge_labels(graph, pos, edge_labels=edge_labels, font_size=8, ax=ax)
        ax.axis('off')

    ani = animation.FuncAnimation(fig, update, frames=total_frames, repeat=False, interval=1200)

    print(f"{filename} 저장 중...")
    try:
        ani.save(filename, writer='ffmpeg', fps=1)
        print(f"성공적으로 저장되었습니다: {filename}")
    except Exception as e:
        print(f"mp4 저장 실패: {e}. 대안으로 GIF를 시도합니다.")
        ani.save(filename.replace('.mp4', '.gif'), writer='pillow', fps=1)
    plt.close(fig)

# ==========================================
# 5. 실행 및 파일 저장
# ==========================================

plt.figure(figsize=(12, 8))
plt.title("서울시 음식점 이동 시간 그래프 (기본 형태)", fontsize=16, fontweight='bold')
edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw(G, pos, with_labels=True, node_color='skyblue', node_size=700, font_family='NanumBarunGothic', font_size=10, edge_color='gray', alpha=0.7)
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=8)
plt.show()

# 2) DFS 애니메이션 저장
build_animation(G, pos, dfs_order, 'dfs.mp4', 'DFS (깊이 우선 탐색)')

# 3) BFS 애니메이션 저장
build_animation(G, pos, bfs_order, 'bfs.mp4', 'BFS (너비 우선 탐색)')

# 4) 프림 알고리즘 애니메이션 저장
build_mst_animation(G, pos, prim_history, 'prim.mp4', '프림 (Prim) 알고리즘')

# 5) 크루스칼 알고리즘 애니메이션 저장
build_mst_animation(G, pos, kruskal_history, 'kruskal.mp4', '크루스칼 (Kruskal) 알고리즘')
```
